```python
def roi_pool(feature_map, roi, output_size=(7, 7)):
    """Extract fixed-size features from arbitrary-sized RoI"""

    x1, y1, x2, y2 = roi
    h, w = y2 - y1, x2 - x1

    oh, ow = output_size
    cell_h = h / oh
    cell_w = w / ow

    out = np.zeros(output_size)

    for i in range(oh):
        for j in range(ow):

            r1 = int(y1 + i * cell_h)
            r2 = int(y1 + (i + 1) * cell_h)

            c1 = int(x1 + j * cell_w)
            c2 = int(x1 + (j + 1) * cell_w)

            r2 = max(r2, r1 + 1)
            c2 = max(c2, c1 + 1)

            r2 = min(r2, feature_map.shape[0])
            c2 = min(c2, feature_map.shape[1])

            if r1 < r2 and c1 < c2:
                out[i, j] = feature_map[r1:r2, c1:c2].max()

    return out


# Simulate feature map from backbone CNN
np.random.seed(42)
feat_map = np.random.rand(50, 50)  # H×W feature map

# Regions of interest
rois = [
    [5, 5, 25, 20],     # Small RoI
    [10, 15, 40, 45],   # Large RoI
    [30, 5, 50, 30],    # Another RoI
]

print("=== ROI POOLING ===")
print("Key insight: Map image coordinates to feature map coordinates")
print("  feature_coord = image_coord / stride")
print("  (stride = downscaling factor of backbone, e.g., 16 for VGG)")
print()

print(f"Feature map shape: {feat_map.shape}")
print("Output size: 7x7 (fixed regardless of RoI size)")
print()

pooled = [
    roi_pool(feat_map, roi, output_size=(7, 7))
    for roi in rois
]

for i, (roi, p) in enumerate(zip(rois, pooled)):
    roi_w = roi[2] - roi[0]
    roi_h = roi[3] - roi[1]

    print(
        f'  RoI {i+1}: {roi} '
        f'(size {roi_w}x{roi_h}) -> pooled {p.shape}'
    )

    print(
        f'           mean={p.mean():.3f}, '
        f'std={p.std():.3f}'
    )
```
